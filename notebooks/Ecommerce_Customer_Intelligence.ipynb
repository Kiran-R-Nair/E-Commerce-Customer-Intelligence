## 1. Problem Statement

E-commerce businesses capture extensive information across customer interactions, website sessions, product engagement, cart activity, transactions, and purchases. However, when these behavioral and transactional signals are analyzed in isolation, the complete customer journey from initial engagement to purchase remains difficult to understand. As a result, businesses may not be able to clearly determine where customers disengage, which behaviors distinguish purchasing from non-purchasing users, or which product, device, engagement, and promotional factors are associated with conversion. This creates a gap between the large volume of customer data available and the ability to turn that data into timely, measurable decision support. The project addresses this gap by developing a unified analytical framework for understanding customer behavior and conversion patterns and by extending the analysis with a purchase/conversion prediction model based only on information available before the purchase outcome.

## 2. Project Objectives

The primary objective of this project is to transform e-commerce customer, website, product, transaction, and marketing data into actionable insights that can support customer understanding, conversion analysis, and purchase-related decision-making.

The project objectives are to:

1. **Understand the E-Commerce Data**
   - Examine the structure, quality, relationships, and business meaning of the available customer, website, product, order, inventory, purchase, and marketing data.

2. **Analyze Customer Behavior**
   - Identify meaningful customer segments and behavioral patterns based on engagement, browsing, purchasing, and transaction activity.

3. **Analyze the Customer Conversion Journey**
   - Examine the progression from website engagement and product interaction to cart activity and completed purchases to identify important conversion patterns and potential drop-off points.

4. **Identify Key Drivers of Conversion**
   - Determine which customer, behavioral, product, device, engagement, and promotional factors are associated with higher or lower purchase conversion.

5. **Develop a Purchase/Conversion Prediction Model**
   - Build and evaluate a machine learning model to estimate purchase/conversion likelihood using only information that would be available before the purchase outcome, thereby reducing the risk of data leakage.

6. **Translate Analytical Findings into Business Recommendations**
   - Convert the analytical and predictive findings into actionable recommendations for customer targeting, conversion improvement, engagement strategies, and e-commerce decision-making.

## 4.5 Outlier Screening

Outlier analysis is performed to identify unusually high or low observations
that may represent extreme customer behavior or potential data-quality issues.

For the purchase prediction dataset:

- `product_views` is a numerical count variable and will be checked for
  potential outliers using the IQR method.
- `add_to_cart` and `begin_checkout` are binary behavioral variables, so
  conventional numerical outlier detection is not applicable.
- `purchased` is the binary target variable and is not treated as an
  outlier-analysis variable.

Potential outliers will be investigated rather than automatically removed,
because unusually high website engagement may represent genuine high-intent
customer behavior and could contain useful predictive information.

```python
# Identify potential outliers in the continuous numerical feature
# using the IQR (Interquartile Range) method.

Q1 = df["product_views"].quantile(0.25)
Q3 = df["product_views"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

product_views_outliers = df[
    (df["product_views"] < lower_bound) |
    (df["product_views"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:", len(product_views_outliers))
print(
    "Potential Outlier Percentage:",
    round(len(product_views_outliers) / len(df) * 100, 2),
    "%"
)
```

```python
# Display the range of product views among potential outliers

product_views_outliers["product_views"].describe()
```

### Outlier Analysis Conclusion

The IQR method identified 439 potential outliers in `product_views`,
representing 1.65% of the dataset.

The identified observations have between 8 and 13 product views. These values
are considered plausible levels of high customer engagement rather than
obvious data-quality errors.

Because high product-view activity may represent genuine high-intent customer
behavior and can provide useful predictive information, these observations
were retained.

No outlier removal was performed for the ML dataset.

### 5.2.4 Traffic Source vs. Purchase

We compare purchase rates across traffic sources.

This helps determine whether visitors arriving from different acquisition
channels have different conversion behavior.

```python
# Calculate sessions and purchase rate by traffic source
traffic_purchase = (
    df.groupby("traffic_source")["purchased"]
      .agg(
          sessions="count",
          purchases="sum",
          purchase_rate="mean"
      )
)

# Convert purchase rate to percentage
traffic_purchase["purchase_rate"] = (
    traffic_purchase["purchase_rate"] * 100
).round(2)

traffic_purchase
```

```python
# Visualize purchase rate by traffic source
plt.figure(figsize=(9, 5))

sns.barplot(
    data=traffic_purchase.reset_index(),
    x="traffic_source",
    y="purchase_rate"
)

plt.title("Purchase Rate by Traffic Source")
plt.xlabel("Traffic Source")
plt.ylabel("Purchase Rate (%)")

plt.xticks(rotation=45)

plt.show()
```

## 7.3 One-Hot Encoding and Feature Scaling

The dataset contains both numerical and categorical variables, so
different preprocessing techniques are required.

### Categorical Features

Categorical features cannot be directly used by Logistic Regression
because they contain text values.

We will use One-Hot Encoding to convert each category into a binary
numerical feature.

### Numerical Features

Numerical features will be standardized using StandardScaler.

Standardization places numerical variables on a comparable scale,
which is useful for Logistic Regression.

ColumnTransformer will be used to apply the appropriate transformation
to each feature group.

The preprocessing pipeline will be fitted only on the training data
to prevent data leakage.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Create the preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        
        # Standardize numerical features
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        
        # One-Hot Encode categorical features
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate

categorical_features = [
    "traffic_source",
    "campaign_name",
    "device_category",
    "city"
]

numerical_features = [
    "product_views",
    "add_to_cart",
    "begin_checkout",
    "year",
    "month",
    "day_of_week"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ]
)
```

## 7.4 Fit and Transform the Data

The preprocessing pipeline is fitted using only the training data.

This prevents information from the test dataset from influencing the
scaling and encoding process.

The training data is fitted and transformed using `fit_transform()`.

The testing data is transformed using `transform()` with the rules
learned from the training data.

```python
# Fit the preprocessing pipeline on the training data
# and transform the training features.
X_train_processed = preprocessor.fit_transform(X_train)

# Transform the testing features using the fitted pipeline.
X_test_processed = preprocessor.transform(X_test)
```

Logistic Regression is a supervised machine learning algorithm used for

## 9.2 Six-Model Stratified Cross-Validation

To obtain a more reliable estimate of model performance, all six classification models are evaluated using five-fold Stratified Cross-Validation.

Stratification is used to preserve the class distribution of the purchase target in every fold.

The test set is not used for model selection. It will remain completely unseen until the final evaluation of the selected model.

The models are compared using:

- Accuracy
- Precision
- Recall
- F1 Score
- ROC-AUC

Because the purchase prediction problem is class-imbalanced, particular attention is given to Recall, Precision, F1 Score, and ROC-AUC rather than Accuracy alone.

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}
```

```python
from sklearn.pipeline import Pipeline

cv_models = {}

for name, model in models.items():

    cv_models[name] = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_results = []

for name, pipeline in cv_models.items():

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Accuracy Mean": scores["test_accuracy"].mean(),
        "Accuracy Std": scores["test_accuracy"].std(),
        "Precision Mean": scores["test_precision"].mean(),
        "Recall Mean": scores["test_recall"].mean(),
        "F1 Mean": scores["test_f1"].mean(),
        "ROC-AUC Mean": scores["test_roc_auc"].mean()
    })

cv_comparison = (
    pd.DataFrame(cv_results)
    .sort_values("ROC-AUC Mean", ascending=False)
    .reset_index(drop=True)
)

cv_comparison
```

```python
# Outlier Analysis — Customer & Order Analytics

outlier_columns = [
    "total_orders",
    "total_revenue",
    "total_net_value",
    "average_order_value",
    "total_discount",
    "total_items",
    "returned_items"
]

outlier_summary = []

for col in outlier_columns:

    Q1 = customer_summary[col].quantile(0.25)
    Q3 = customer_summary[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = customer_summary[
        (customer_summary[col] < lower_bound) |
        (customer_summary[col] > upper_bound)
    ]

    outlier_summary.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier %": round(
            len(outliers) / len(customer_summary) * 100, 2
        )
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

outlier_summary_df
```

### 13.6 Outlier Analysis

The IQR (Interquartile Range) method was used to identify potential outliers in key inventory and supply variables, including stock levels, sales velocity, inventory coverage, order quantities, unit cost, and lead time.

Potential outliers were not automatically removed because extreme inventory values can represent genuine business conditions, such as high-demand products, large stock holdings, bulk purchase orders, higher procurement costs, or longer supplier lead times.

Therefore, the detected outliers will be treated as potential business observations rather than data errors unless further investigation indicates otherwise.

```python
inventory_outlier_columns = [
    "units_in_stock",
    "units_sold_7d",
    "units_sold_30d",
    "units_sold_60d",
    "days_of_inventory_left",
    "total_order_quantity",
    "avg_cost_per_unit",
    "avg_lead_time"
]

inventory_outliers = []

for col in inventory_outlier_columns:

    Q1 = sku_inventory_supply[col].quantile(0.25)
    Q3 = sku_inventory_supply[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (sku_inventory_supply[col] < lower_bound) |
        (sku_inventory_supply[col] > upper_bound)
    ).sum()

    inventory_outliers.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier %": round(
            outlier_count / len(sku_inventory_supply) * 100, 2
        )
    })

inventory_outlier_summary = pd.DataFrame(inventory_outliers)

inventory_outlier_summary
```

### 13.6.1 Outlier Interpretation

The IQR method identified potential outliers in several inventory and supply variables. The highest potential outlier percentages were observed for `total_order_quantity`, `avg_cost_per_unit`, `avg_lead_time`, and `days_of_inventory_left`.

However, these observations were not automatically removed. Several variables are highly right-skewed, indicating that extreme values may represent genuine inventory and supply-chain behavior rather than data errors.

In particular, `total_order_quantity`, `avg_cost_per_unit`, and `avg_lead_time` have a first quartile and third quartile of zero. Consequently, the IQR method flags non-zero observations as potential outliers because the majority of observations are zero. These should therefore not be interpreted as data-quality errors solely based on the IQR rule.

The identified extreme values are retained for business analysis because they may represent high-demand products, procurement activity, higher-cost products, or longer supplier lead times.

### 14.4 Outlier Analysis

The IQR method was used to identify potential outliers in website engagement, conversion, revenue, and marketing performance variables.

The analysis considers extreme observations as potential business observations rather than automatically treating them as data errors. High session volumes, product views, purchases, revenue, advertising spend, impressions, or conversion values may represent genuine differences in traffic or campaign performance.

Skewness was also examined to determine whether the numerical variables were substantially concentrated toward lower values with a smaller number of high-value observations.

Potential outliers were therefore retained for further analysis unless there was evidence that they represented invalid or erroneous records.

```python
marketing_outlier_columns = [
    "sessions",
    "product_views",
    "add_to_cart",
    "begin_checkout",
    "purchases",
    "revenue",
    "conversion_rate",
    "aov",
    "meta_spend",
    "meta_reach",
    "meta_impressions",
    "meta_link_clicks",
    "meta_add_to_cart",
    "meta_checkouts",
    "meta_purchases",
    "meta_conversion_value"
]

marketing_outliers = []

for col in marketing_outlier_columns:

    Q1 = marketing[col].quantile(0.25)
    Q3 = marketing[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (marketing[col] < lower_bound) |
        (marketing[col] > upper_bound)
    ).sum()

    marketing_outliers.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier %": round(
            outlier_count / len(marketing) * 100, 2
        )
    })

marketing_outlier_summary = pd.DataFrame(marketing_outliers)

marketing_outlier_summary
```

```python
marketing_skewness = (
    marketing[marketing_outlier_columns]
    .skew()
    .sort_values(ascending=False)
    .to_frame("Skewness")
)

marketing_skewness
```

### 14.4.1 Outlier Interpretation

The IQR analysis identified potential outliers across website engagement, conversion, revenue, and marketing variables.

The potential outliers were retained because extreme values may represent genuine high-traffic sessions, high-performing campaigns, or unusually strong sales activity rather than data errors.

`begin_checkout` and `add_to_cart` showed relatively high outlier percentages because most observations contain zero or very low funnel activity. Therefore, the IQR method flags normal positive activity as statistical outliers in these sparse variables.

The Meta advertising variables showed very high positive skewness because most observations have zero Meta advertising activity, while a small number of observations contain substantial advertising values.

No automatic outlier removal was performed because these observations may contain important business information.

### Business Implication

Extreme observations should be investigated as potentially high-performing traffic, conversion, or marketing opportunities rather than automatically treated as errors. The strong right-skewness of several marketing variables also indicates that median-based measures and segmented analysis may be more informative than relying only on averages.
